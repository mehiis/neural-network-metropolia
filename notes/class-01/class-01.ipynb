```python
(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

#vedetään yhdeksi vektoriksi
#verkot tykkää pienistä luvuista
x_train = x_train.reshape( (60000, 784) ).astype('float32') / 255
x_test = x_test.reshape( (10000, 784) ).astype('float32') / 255
```

```python
#rakennetaan verkkoa, siinä on yleensä kolme vaihetta
#layerin on vähän niin kuin funktioita

#yleisesti3 vaihetta

#1. Luodaan malli (Create): Kerrotaan monta solua syötteessä (input)
inputs = keras.Input(shape=(784,)) #sisältää

x0 = keras.layers.Dense(64, activation='relu')(inputs) #Dense on täysin kytketty
x1 = keras.layers.Dense(32, activation='relu')(x0)

outputs = keras.layers.Dense(10, activation='softmax')(x1)

model = keras.Model(inputs=inputs, output=outputs, name='mnist_model')

model.summary()

#2. Compile
model.compile(
    loss = keras.losses.SparseCategorycalCrossentropy(),
    optimizer = keras.optimizers.Adam(),
    metrics = ['accuracy'], 
)
    

#3. Sovitus/Koulutus(Model fit/Train): Saattaa kestää pitkäänkin, riippuen datan määrästä

#batch_size laskee 64 kuvan keskiarvon ja päivittää mallia, eli ei katso yksi kuva kerrallaan.
#epochs käy 20 kertaa kaikki kuvat läpi.
#validation_split tarkoittaa koulutusmäärän kokoa, jokaisen epochsin jälkeen tulee testi!
history = model.fit(x_train, y_train, batch_size=64, epochs=20, validation_split=0.2) 
plt.plot(history.history['loss'])
plt.plot(history.history['val_loss'])
plt.grid()
plt.show() #näkyy kannattiko kouluttaa 20 kierrosta. (sininen = training loss, ja oranssi = 'validaatio loss')

#lasketaan tarkkuus
test_scores = model.evaluat(x_test, y_test, verbose=2)

#loss = 
#accuracy = 
#validation = 20% testattu accuracy
#jos arvot alkaa paranemaan jatka kouluttamista, jos alkaa huononemaan ei tarvitse jatkaa koulutusta

#4. Käyttö(Evaluate/Predict): Testaa ja käytä dataa

```
