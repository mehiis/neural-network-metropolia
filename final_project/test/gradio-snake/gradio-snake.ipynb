```python
from pathlib import Path
import gradio as gr

# Polku pelisivulle
game_file = Path.cwd() / "game.html"

# Koko näytön iframe, joka lataa pelin
game_iframe = f"""
<iframe
    id="snake-frame"
    src="/gradio_api/file={game_file.as_posix()}"
    allow="microphone"
    style="position: fixed; inset: 0; width: 100vw; height: 100vh; border: none;">
</iframe>
"""

# Gradio-sovellus
with gr.Blocks(title="Snake") as demo:
    gr.HTML(game_iframe)

# Vapauta mahdolliset vanhat Gradio-instanssit ja käynnistä
gr.close_all()
demo.launch(allowed_paths=[str(game_file.parent)])#INSTALL GRADIO

#https://github.com/gradio-app/gradio
#https://huggingface.co/learn/llm-course/chapter9/1
pip install gradio
```
